```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')
```

```python
print(credits.shape)
movies.shape
```

```python
credits.head(1)['cast'][0][2:12]
```

```python
movies = movies.merge(credits, on ='title')
movies.columns
```

```python
# fetching only important columns
# genres , id , keywords ,overview ,cast_x , crew
movies = movies[['title','genres','id','keywords','overview', 'cast','crew']]
movies.head(1)  
```

```python
# Removing duplicated and null data
print(movies.isnull().sum())
movies.dropna(inplace =True)
print(movies.isnull().sum())
```

```python
movies.duplicated().sum()
```

```python
# Fetching only 'names' from 'genre' and 'keywords' column
import ast
def convert(obj):
    l=[]
    for i in ast.literal_eval(obj):
        l.append(i['name'])
    return l

movies['genres'] = movies['genres'].apply(convert)
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

```python
print(movies['cast'].head(2))
# Fetching only 'names' from 'cast' column(will be excluding name of only 3 characters to include in the data)
def convert_cast(obj):
    l=[]
    count =0
    for i in ast.literal_eval(obj):
        if count !=3:
          l.append(i['name'])
          count +=1
        else:
            break
    return l

movies['cast'] = movies['cast'].apply(convert_cast) 
movies.head()
```

```python
movies['cast']
```

```python
print(movies['crew'].head(2))
# Fetching Director's name from crew

def fetch_director(obj):
    l=[]
    for i in ast.literal_eval(obj):
           if i['job']=='Director':
               l.append(i['name'])
               break
    return l

movies['crew'] = movies['crew'].apply(fetch_director)
movies.head()
```

```python
# Converting string into list
movies['overview'] = movies['overview'].apply(lambda x : x.split())
```

```python
# Removing spaces to create unique tags
movies['overview'] = movies['overview'].apply(lambda x : [i.replace(' ','') for i in x])
movies['genres'] = movies['genres'].apply(lambda x : [i.replace(' ','') for i in x])
movies['cast'] = movies['cast'].apply(lambda x : [i.replace(' ','') for i in x])
movies['crew'] = movies['crew'].apply(lambda x : [i.replace(' ','') for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x : [i.replace(' ','') for i in x])
```

```python
movies.head()
```

```python
# Creating a single tag column from title,genres,keywords,overview,cast ,crew
movies['tags'] = movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
movies.head()
```

```python
# Creating a new Dataset
new_df = movies[['id','title','tags']]
new_df.head()
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:' '.join(x) )
new_df.head()
```

```python
new_df['tags'][0]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:x.lower() )
new_df.head()
```

```python
#!pip install scikit-learn
#from sklearn import feature_extraction
from sklearn.feature_extraction.text import CountVectorizer       
cv = CountVectorizer(max_features=5000, stop_words='english')   
vector = cv.fit_transform(new_df['tags']).toarray()
print(cv.get_feature_names_out())
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    l=[]
    for i in text.split():
        l.append(ps.stem(i))
    return ' '.join(l)    

new_df['tags'] = new_df['tags'].apply(stem) 
cv.fit_transform(new_df['tags']).toarray()
```

```python
cv.get_feature_names_out()
```

```python
vector.shape
```

```python
# Finding similarty between vectors
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
```

```python
# Creating function for recommendation
def recommend(movie):
    movie_index = new_df[new_df['title']==movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)) , reverse = True , key = lambda x:x[1])[1:6]
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
        
```

```python
recommend('Avatar')
```

```python
import pickle
pickle.dump(new_df, open("movies.pkl", "wb"))
pickle.dump(similarity, open("similarity.pkl", "wb"))
pickle.dump(cv, open("vectorizer.pkl", "wb"))
```
